```python
import matplotlib.pyplot as plt
import csv
from scipy import stats
import numpy as np
import sklearn as sl
from sklearn import linear_model
```

```python
# Load the provided data file with the used car data (you can also have a look at it with any text editor)

filename = "data/km_year_power_price.csv"
lines = csv.reader(open(filename, newline=''), delimiter=',')

  
dataset = list(lines)
# print(dataset)

# convert to numpy array:
npdata = np.array(dataset)
print(npdata)
print("dataset has this many entries:", len(dataset))
# place your loading code here
```

Use linear regression to estimate the car prices from the year, kilometers or engine power. 
You can make a simple 1D regression from each one of the parameters independently 




```python
# linear regression with linregress (estimate price from year)
# first get the columns i need:

prices = np.array([float(x) for x in npdata[1:,3]])
years = np.array([float(x) for x in npdata[1:,1]])

# print(prices)
# print(years)

result1 = stats.linregress(years,prices)
m1 = result1[0]
q1 = result1[1] 

# let's see results
print(result1) 
for i in result1:
    print(i)

# your code.....
```

```python
# Plot the data and the lines representing the output of the linregress and least square algorithms
plt.plot(years, prices, 'o', label='original data')
plt.plot(years, q1 + m1*years, 'r', label='fitted line')
# plt.plot(years, q2 + m2*years, 'r', label='fitted line')
# plt.plot(years, q3 + m3*years, 'r', label='fitted line')
plt.legend()
plt.show()
```

```python
# linear regression with linregress (estimate price from power)
powers = np.array([float(x) for x in npdata[1:,2]])
result2 = stats.linregress(powers,prices)
m2 = result2[0]
q2 = result2[1] 
```

```python
# linear regression with linregress (estimate price from km)
kms = np.array([float(x) for x in npdata[1:,0]])
result3 = stats.linregress(kms,prices)
m3 = result3[0]
q3 = result3[1] 
```

```python
# Plot the data and the lines representing the output of the linregress and least square algorithms
plt.plot(years, prices, 'o', label='original data')
plt.plot(years, q1 + m1*years, 'r', label='fitted line')
#plt.plot(powers, q2 + m2*powers, 'g', label='fitted line')
#plt.plot(kms, q3 + m3*kms, 'y', label='fitted line')
plt.legend()
plt.show()
```
